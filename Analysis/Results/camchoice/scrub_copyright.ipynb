```python
import pandas as pd
from pathlib import Path
import shutil
```

```python
extension = ".parquet"  # Change to your desired extension

cols_to_delete = [
    'Question',
    'Answers',
    'Answer_Ratios',
    'Correct_Answer',
    'formatted_question_mcqa',
]

archive = Path("./unscrubbed_archive")
archive.mkdir(exist_ok=True)

for filepath in Path(".").glob(f"*{extension}"):
    shutil.copy2(filepath, archive / filepath.name)
    df = pd.read_parquet(filepath)
    df.drop(columns=cols_to_delete, errors='ignore', inplace=True)
    df.to_parquet(filepath)
    print(f'Successfully scrubbed file: {filepath}')
```
